# Warehouse data (linked with Hearing Australia audiometry)

```python
patron_hl_data = pd.read_csv('/Users/mcab8311/Documents/HEIDI/audiogram_HL_demographics.csv')

plt.figure(figsize=(10, 5))
sn.histplot(data=patron_hl_data, x='HL_AGE', hue='GENDER_TEXT', kde=True, binwidth=2, stat='percent')
plt.title('Demographics of people with hearing loss (HEIDI warehouse)')
patron_hl_data['GENDER_TEXT'] = patron_hl_data['GENDER_TEXT'].astype('category')
```

```python
# frank
frank_patron_hl_data = pd.read_csv('/Users/z5116060/Desktop/GitProjects/hearing/ftran/data/frank_audiogram.csv')
sampled_data = frank_patron_hl_data.sample(frac=0.1, random_state=1)

sn.scatterplot(data=sampled_data, y='INITIAL_3FAHL_RIGHT', x='HL_AGE', hue='GENDER_TEXT')

# Regression lines for each group
groups = sampled_data['GENDER_TEXT'].unique()
for group in groups:
    sn.regplot(y='INITIAL_3FAHL_RIGHT', x='HL_AGE', data=sampled_data[sampled_data['GENDER_TEXT'] == group], scatter=False, label=f'Group {group}')

plt.figure(figsize=(5, 5))
plt.legend(title='Gender')
```
